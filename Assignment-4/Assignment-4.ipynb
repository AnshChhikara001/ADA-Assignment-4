```python
# Task 1: Backtracking (N-Queens)

def is_safe(board, row, col, n):
    
    # Check column
    for i in range(row):
        if board[i][col] == 1:
            return False
    
    # Check left diagonal
    i, j = row-1, col-1
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1
    
    # Check right diagonal
    i, j = row-1, col+1
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1
    
    return True


def solve_n_queens(board, row, n):
    if row == n:
        return True
    
    for col in range(n):
        if is_safe(board, row, col, n):
            board[row][col] = 1
            
            if solve_n_queens(board, row+1, n):
                return True
            
            board[row][col] = 0   # backtrack
    
    return False


# Run
n = 4
board = [[0]*n for _ in range(n)]

if solve_n_queens(board, 0, n):
    print("Solution:")
    for row in board:
        print(row)
else:
    print("No solution")
```

```python
# Task 2: Branch and Bound (0/1 Knapsack)

class Item:
    def __init__(self, value, weight):
        self.value = value
        self.weight = weight

# Calculate upper bound
def bound(node_level, profit, weight, capacity, items):
    if weight >= capacity:
        return 0
    
    bound_profit = profit
    total_weight = weight
    n = len(items)
    
    i = node_level + 1
    
    while i < n and total_weight + items[i].weight <= capacity:
        total_weight += items[i].weight
        bound_profit += items[i].value
        i += 1
    
    # Take fraction (approximation)
    if i < n:
        bound_profit += (capacity - total_weight) * (items[i].value / items[i].weight)
    
    return bound_profit


def knapsack_branch_bound(items, capacity):
    items.sort(key=lambda x: x.value/x.weight, reverse=True)
    
    max_profit = 0
    
    def dfs(level, profit, weight):
        nonlocal max_profit
        
        if weight <= capacity and profit > max_profit:
            max_profit = profit
        
        if level == len(items) - 1:
            return
        
        if bound(level, profit, weight, capacity, items) < max_profit:
            return   # PRUNE
        
        # Include item
        dfs(level + 1, profit + items[level+1].value, weight + items[level+1].weight)
        
        # Exclude item
        dfs(level + 1, profit, weight)
    
    dfs(-1, 0, 0)
    return max_profit


# Input
items = [
    Item(60, 10),
    Item(100, 20),
    Item(120, 30)
]

capacity = 50

# Run
result = knapsack_branch_bound(items, capacity)

print("Maximum Profit:", result)
```

```python
# Task 3:

def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    
    comparisons = 0
    
    for i in range(n - m + 1):
        match = True
        
        for j in range(m):
            comparisons += 1
            if text[i + j] != pattern[j]:
                match = False
                break
        
        if match:
            print("Pattern found at index:", i)
    
    return comparisons
```

```python
def compute_lps(pattern):
    lps = [0] * len(pattern)
    length = 0
    i = 1
    
    while i < len(pattern):
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1
    
    return lps


def kmp_search(text, pattern):
    n = len(text)
    m = len(pattern)
    
    lps = compute_lps(pattern)
    
    i = j = 0
    comparisons = 0
    
    while i < n:
        comparisons += 1
        
        if text[i] == pattern[j]:
            i += 1
            j += 1
        
        if j == m:
            print("Pattern found at index:", i - j)
            j = lps[j - 1]
        
        elif i < n and text[i] != pattern[j]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    
    return comparisons
```

```python
text = "ABABDABACDABABCABAB"
pattern = "ABABCABAB"

print("Naive Search:")
naive_comp = naive_search(text, pattern)
print("Comparisons:", naive_comp)

print("\nKMP Search:")
kmp_comp = kmp_search(text, pattern)
print("Comparisons:", kmp_comp)
```

```python
# Task 4:

import time

texts = [
    "A" * 50 + "B",
    "A" * 100 + "B",
    "A" * 200 + "B"
]

pattern = "AB"

print("Comparison of Naive vs KMP:\n")

for text in texts:
    
    print(f"Text length: {len(text)}")
    
    # Naive
    start = time.time()
    naive_comp = naive_search(text, pattern)
    naive_time = time.time() - start
    
    # KMP
    start = time.time()
    kmp_comp = kmp_search(text, pattern)
    kmp_time = time.time() - start
    
    print(f"Naive -> Comparisons: {naive_comp}, Time: {naive_time}")
    print(f"KMP   -> Comparisons: {kmp_comp}, Time: {kmp_time}")
    
    print("-" * 40)
```

```python
# Task 5: Conceptual Programming with Comments
# Problem: N-Queens using Backtracking

def is_safe(board, row, col, n):
    # Check if any queen is placed in same column
    for i in range(row):
        if board[i][col] == 1:
            return False
    
    # Check left diagonal
    i, j = row-1, col-1
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1
    
    # Check right diagonal
    i, j = row-1, col+1
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1
    
    return True


def solve_n_queens(board, row, n):
    # Base case: all queens placed
    if row == n:
        return True
    
    # Try placing queen in each column
    for col in range(n):
        
        # Check if safe to place queen
        if is_safe(board, row, col, n):
            
            # Place queen
            board[row][col] = 1
            
            # Recur for next row
            if solve_n_queens(board, row+1, n):
                return True
            
            # Backtrack: remove queen
            board[row][col] = 0
    
    return False


# Initialize board
n = 4
board = [[0]*n for _ in range(n)]

# Solve problem
if solve_n_queens(board, 0, n):
    print("Solution:")
    for row in board:
        print(row)
else:
    print("No solution found")
```
